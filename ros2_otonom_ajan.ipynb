```python
from google import genai
from google.colab import userdata

# 1. Colab Secrets menüsünden API anahtarını güvenli şekilde al
api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)

# 2. Az önce Colab sisteminde oluşturduğumuz ajan dosyasını oku
with open("ros2-navigation-specialist.md", "r", encoding="utf-8") as f:
    system_prompt = f.read()

# 3. Ajana sorulacak soru (Kendi ROS 2 sorunu buraya yazabilirsin)
soru = "ROS 2 projemde bir USB kameram var. Bu kameradan gelen anlık görüntüyü cv_bridge ile alıp, PyTorch kullanarak 'kırmızı renkli engelleri' tespit eden basit bir Python node'u yazar mısın?"

# 4. Modele ajanın kimliğini sistem talimatı olarak ver ve yanıtı al
print("ROS 2 Uzmanı Ajan düşünüyor...\n")
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=soru,
    config={"system_instruction": system_prompt}
)

print("--- Ajan Yanıtı ---\n")
print(response.text)
```
