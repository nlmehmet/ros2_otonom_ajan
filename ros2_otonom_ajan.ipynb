```python
from google import genai
from google.colab import userdata

api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)

# 1. Her iki ajanın kimlik dosyalarını okuyalım
with open("inventor-cad-automation-specialist.md", "r", encoding="utf-8") as f:
    cad_system_prompt = f.read()

with open("ros2-navigation-specialist.md", "r", encoding="utf-8") as f:
    ros2_system_prompt = f.read()

# --- BİRİNCİ AŞAMA: CAD AJANI ÇALIŞIYOR ---
print("⚙️ ADIM 1: CAD Ajanı mekanik tasarımı yapıyor...\n")
cad_soru = "Bir kamera için 3 eksenli pan-tilt-roll gimbal mekanizması tasarlıyorum. Pan, tilt ve roll eksenlerinin kinematik bağlantı noktalarını ve aralarındaki mafsal (joint) hiyerarşisini Python Inventor API ile nasıl kuracağımı özetleyen kısa bir teknik rapor ver."

cad_response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=cad_soru,
    config={"system_instruction": cad_system_prompt}
)
cad_ciktisi = cad_response.text
print("✅ CAD Tasarımı Tamamlandı!\n")


# --- İKİNCİ AŞAMA: ROS 2 AJANI ÇALIŞIYOR ---
print("🤖 ADIM 2: ROS 2 Ajanı bu tasarımı alıp yazılıma döküyor...\n")

# CAD ajanından gelen çıktıyı ROS 2 ajanına bir "görev brifingi" olarak veriyoruz
ros2_soru = f"""
Mekanik tasarım departmanımız aşağıdaki pan-tilt-roll gimbal sistemini tasarladı:

{cad_ciktisi}

Lütfen yukarıdaki mekanik kısıtları ve hiyerarşiyi okuyarak, bu 3 ekseni (pan, tilt, roll) kontrol edecek basit bir ROS 2 Python node'u (sensor_msgs/JointState kullanarak) yaz.
"""

ros2_response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=ros2_soru,
    config={"system_instruction": ros2_system_prompt}
)

# --- SONUÇLARI EKRANA BAS ---
print("===========================================")
print("🎯 NİHAİ SİSTEM ÇIKTISI (ROS 2 AJANI)")
print("===========================================\n")
print(ros2_response.text)
```
